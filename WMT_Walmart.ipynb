```python

import yfinance as yf
import pandas as pd
import numpy as np
#setup
#code
ticker = yf.Ticker('WMT') #import data
aapl_df = ticker.history(period="5y") #get data from 5 year period in dataframe
aapl_df.drop(['High','Close','Volume','Dividends','Stock Splits'], axis=1, inplace=True) #only have two columns, not seven
open = np.empty(shape = (1259), dtype = float)
close = np.empty(shape = (1259), dtype = float)

open=aapl_df[['Open']].to_numpy()
```

```python
print(open)
```

```python
X = np.zeros((1260,3))
Y = [0]*1260
for i in range(1250):
 X[i] = [open[i][0],open[i+1][0],open[i+2][0]]
 Y[i] = open[i+3][0]
print(X[0])
print(Y[0])
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(
    X, Y, test_size=0.33, random_state=42)
```

4-5

```python
print(X_train[0])
print(y_train[0])
```

```python
from sklearn import linear_model
from sklearn.metrics import mean_squared_error, r2_score
linr = linear_model.LinearRegression()
linr.fit(X_train, y_train)
y_pred_linear_regression = linr.predict(X_test)
y_pred_linr_train = linr.predict(X_train)
# The coefficients
print("Coefficients: \n", linr.coef_)
# The mean squared error
print("Mean squared error: %.2f" % mean_squared_error(y_test, y_pred_linear_regression))
inverse_linr = 1/mean_squared_error(y_test, y_pred_linear_regression)
inverse_linr_train = 1/mean_squared_error(y_train, y_pred_linr_train)
```

```python
from sklearn.neural_network import MLPRegressor
regr = MLPRegressor(max_iter=2000, tol=1e-4)
regr.fit(X_train, y_train)
y_pred_mlp = regr.predict(X_test)
y_pred_mlp_train = regr.predict(X_train)
print("Mean squared error: %.2f" % mean_squared_error(y_test, y_pred_mlp))
inverse_regr = 1/mean_squared_error(y_test, y_pred_mlp)
inverse_regr_train = 1/mean_squared_error(y_train, y_pred_mlp_train)
```

```python
from sklearn.neural_network import MLPRegressor # second set of parameters
regr_two = MLPRegressor(max_iter=4000, tol=1e-6)
regr_two.fit(X_train, y_train)
y_pred_mlp_two = regr_two.predict(X_test)
y_pred_mlp_two_train = regr_two.predict(X_train)
print("Mean squared error: %.2f" % mean_squared_error(y_test, y_pred_mlp_two))
inverse_regr_two = 1/mean_squared_error(y_test, y_pred_mlp_two)
inverse_regr_two_train = 1/mean_squared_error(y_train, y_pred_mlp_train)
```

```python
from sklearn.model_selection import cross_val_score
from sklearn.tree import DecisionTreeRegressor
regressor = DecisionTreeRegressor(max_depth = 7)
regressor.fit(X_train, y_train)
y_pred_decision_tree = regressor.predict(X_test)
y_pred_decision_train = regressor.predict(X_train)
print("Mean squared error: %.2f" % mean_squared_error(y_test, y_pred_decision_tree))
inverse_regressor = 1/mean_squared_error(y_test, y_pred_decision_tree)
inverse_regressor_train = 1/(mean_squared_error(y_train, y_pred_decision_train) + 0.0001)
```

```python
from sklearn.model_selection import cross_val_score # second set of parameters
from sklearn.tree import DecisionTreeRegressor
regressor_two = DecisionTreeRegressor(max_depth = 10)
regressor_two.fit(X_train, y_train)
y_pred_decision_tree_two = regressor_two.predict(X_test)
y_pred_decision_train_two = regressor_two.predict(X_train)
print("Mean squared error: %.2f" % mean_squared_error(y_test, y_pred_decision_tree_two))
inverse_regressor_two = 1/mean_squared_error(y_test, y_pred_decision_tree_two)
inverse_regressor_two_train = 1/(mean_squared_error(y_train, y_pred_decision_train_two) + 0.0001)
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.datasets import make_regression
rand = RandomForestRegressor(max_depth = 10, n_estimators = 10, random_state=0)
rand.fit(X_train, y_train)
y_pred_random_forest = rand.predict(X_test)
y_pred_rand_train = rand.predict(X_train)
print("Mean squared error: %.2f" % mean_squared_error(y_test, y_pred_random_forest))
inverse_rand = 1/mean_squared_error(y_test, y_pred_random_forest)
inverse_rand_train = 1/mean_squared_error(y_train, y_pred_rand_train)
```

```python
from sklearn.ensemble import RandomForestRegressor # second set of parameters
from sklearn.datasets import make_regression
rand_two = RandomForestRegressor(max_depth = 25, n_estimators = 25, random_state=0)
rand_two.fit(X_train, y_train)
y_pred_random_forest_two = rand_two.predict(X_test)
y_pred_rand_train_two = rand_two.predict(X_train)
print("Mean squared error: %.2f" % mean_squared_error(y_test, y_pred_random_forest_two))
inverse_rand_two = 1/mean_squared_error(y_test, y_pred_random_forest_two)
inverse_rand_train_two = 1/mean_squared_error(y_train, y_pred_rand_train_two)
```

```python
from sklearn.neighbors import KNeighborsRegressor
neigh = KNeighborsRegressor(n_neighbors=2)
neigh.fit(X_train, y_train)
y_pred_neighbors = neigh.predict(X_test)
y_pred_neighbors_train = neigh.predict(X_train)
print("Mean squared error: %.2f" % mean_squared_error(y_test, y_pred_neighbors))
inverse_neigh = 1/mean_squared_error(y_test, y_pred_neighbors)
inverse_neigh_train = 1/mean_squared_error(y_train, y_pred_neighbors_train)
```

```python
from sklearn.neighbors import KNeighborsRegressor
neigh_two = KNeighborsRegressor(n_neighbors=3)
neigh_two.fit(X_train, y_train)
y_pred_neighbors_two = neigh_two.predict(X_test)
y_pred_neighbors_train_two = neigh_two.predict(X_train)
print("Mean squared error: %.2f" % mean_squared_error(y_test, y_pred_neighbors_two))
inverse_neigh_two = 1/mean_squared_error(y_test, y_pred_neighbors_two)
inverse_neigh_train_two = 1/mean_squared_error(y_train, y_pred_neighbors_train_two)
```
